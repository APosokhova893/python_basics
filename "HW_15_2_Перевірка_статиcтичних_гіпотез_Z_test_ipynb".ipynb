```python
sample_mean = np.mean(data)
population_mean =20
population_std = np.std(data, ddof=1)
sample_size =len(data)
alpha = 0.05
```

 Формулюємо гіпотези


H₀: μ = 20(середній чек не змінився)
H₁: μ > 20(середній чек став більшим)
Тип тесту: правосторонній (one-tailed)

Чому саме правосторонній тест?

Тому що нам треба зʼясувати,чи вийшло у нас статистично значущо підвищити середній чек.

```python
standard_error = population_std / np.sqrt(sample_size)
print(f"Стандартне відхилення: {population_std:.3f}")
print(f"Розмір вибірки: {sample_size}")
print(f"Стандартна помилка (SE): {standard_error:.3f}")
```

Стандартна помилка становить SE = 0.625 долара. Це означає, що оцінене стандартне відхилення вибіркового середнього чека для вибірок із 100 покупок становить приблизно 0.625 долара.


```python
z_statistic = (sample_mean - population_mean) / standard_error
print(f"z = ({sample_mean} - {population_mean}) / {standard_error:.3f}")
print(f"z = {sample_mean - population_mean} / {standard_error:.3f}")
print(f"z = {z_statistic:.3f}")
```

```python
z_critical = stats.norm.ppf(1 - alpha)
print(f"Критичне значення Z: {z_critical:.3f}")
```

```python
p_value = 1 - stats.norm.cdf(z_statistic)
print(f"p-value = {p_value:.4f}")
```

```python
if z_statistic > z_critical:
    print("Відхиляємо H0")
else:
    print("Не відхиляємо H0")
```

```python
if p_value < alpha:
    print("Відхиляємо H0. Результат статистично значущий.")
else:
    print("Не відхиляємо H0. Результат статистично незначущий.")
```

За рівня значущості 5% ми не маємо достатніх статистичних доказів того, що середній чек покупців у липні 2025 року збільшився порівняно з попереднім значенням 20 доларів.

Виконайте обчислення z-test з використанням бібліотеки statsmodels.

Отримайте z-статистику та р-значення.

Виведіть p-значення та зробіть висновок, чи ми маємо достатньо доказів, аби стверджувати, що середній чек зріс.

Чи зійшлись значення z-статистику та р-значення в цьому завданні з попередніми обчисленнями?

```python
from statsmodels.stats import weightstats as stests
z_stat_sm, p_val_sm = stests.ztest(
    data,
    value=population_mean,
    alternative='larger'
)
print(f"Z-статистика: {z_stat_sm:.3f}")
print(f"P-value: {p_val_sm:.4f}")
```

Отже, на рівні значущості 5% немає достатніх статистичних доказів того, що середній чек покупців збільшився порівняно з $20.
